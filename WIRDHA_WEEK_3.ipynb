### **1. URUTAN PRIORITAS OPERATOR PADA PROGRAM PYTHON**

Urutan prioritas operator (operator precedence) adalah aturan bahasa program Python yang menentukan operator mana yang harus dieksekusi atau dihitung terlebih dahulu disaat ada lebih dari satu operator dalam sebuah ekspresi matematis/logika. Urutannya dimulai dari tertinggi ke terendah dimulai dari tanda kurung (), eksponen (**), unary (+x, -x, ~x), operasi perkalian/pembagian/modulus (*, /, //, %) dan penjumlahan/pengurangan (+, -), geseran bitwise (<<, >>), bitwise AND (&), bitwise XOR (^), serta bitwise OR (|). Pada tingkat bawah, terdapat perbandingan/identitas/keanggotaan (==, !=, >, >=, <, <=, is, is not, in, not in), dan terakhir operator logika dengan urutan not, and, lalu or.
