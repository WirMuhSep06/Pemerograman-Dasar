### **1. URUTAN PRIORITAS OPERATOR PADA PROGRAM PYTHON**

Urutan prioritas operator (operator precedence) adalah aturan bahasa program Python yang menentukan operator mana yang harus dieksekusi atau dihitung terlebih dahulu disaat ada lebih dari satu operator dalam sebuah ekspresi matematis/logika. Urutannya dimulai dari tertinggi ke terendah dimulai dari tanda kurung (), eksponen (**), unary (+x, -x, ~x), operasi perkalian/pembagian/modulus (*, /, //, %) dan penjumlahan/pengurangan (+, -), geseran bitwise (<<, >>), bitwise AND (&), bitwise XOR (^), serta bitwise OR (|). Pada tingkat bawah, terdapat perbandingan/identitas (==, !=, >, >=, <, <=), dan terakhir operator logika dengan urutan not, and, lalu or.

Sebagai contoh pada ekspresi:





```python
10 + 5 * 2
```

output yang dihasilkan dari operasi tersebut adalah 20 bukan 30, hal ini dikarenakan operator perkalian memiliki prioritas yang lebih tinggi dibanding operator penjumlahan, maka sistem program python ini akan langsung mengoperasikan 5 * 2 = 10 terlebih dahulu dan kemudian 10 + 10 = 20

### **2. PERBEDAN OPERATOR RELASIONAL DAN OPERATOR LOGIKA**
Operator Relasional adalah operator yang digunakan untuk membandingkan dua atau lebih nilai/variabel dan biasanya memiliki output Boolean (True atau False). Contoh (>, <, ==, !=, >=, <=).
Operator Logika adalah operator yang digunakan untuk menggabungkan beberapa kondisi untuk menghasilkan Boolean (True atau False). Contoh (AND , OR, NOT).

Contoh program python:





```python
generator_1 = {"tegangan": 390, "suhu": 83}
generator_2 = {"tegangan": 388, "suhu": 79}


generator_1_aman = (380 < generator_1["tegangan"] < 400) and (70 < generator_1["suhu"] < 90)


generator_2_aman = (380 < generator_2["tegangan"] < 400) and (70 < generator_2["suhu"] < 90)

print("Status Generator 1  :", generator_1_aman)
print("Status Generator 2  :", generator_2_aman)
```

Pada kode tersebut, < merupakan operator relasional yang membandingkan nilai tegangan dan suhu dengan batas aman yang ditentukan. Sementara itu, and merupakan operator logika yang menggabungkan kedua kondisi tersebut. Generator dinyatakan aman apabila tegangan berada di antara 380 hingga 400 dan suhu di antara 70 hingga 90, karena kedua generator tersebut telah memenuhi nilai aman maka output akan menampilkan boolean True (aman)

### **3. MENCARI KASUS KESALAHAN AKIBAT OPERATOR PRECEDENCE**

Program Error:

```python
nilai_uts = 80
nilai_uas = 90
nilai_tugas = 70

rata_rata = nilai_uts + nilai_uas + nilai_tugas / 3

print("Rata-rata:", rata_rata)
```

Python mengevaluasi pembagian nilai_tugas / 3 terlebih dahulu karena operator pembagian (/) memiliki operator precedence yang lebih tinggi daripada operator penjumlahan (+).

Akibatnya, program hanya membagi nilai_tugas dengan 3, lalu menjumlahkan hasilnya dengan nilai_uts dan nilai_uas, sehingga menghasilkan nilai rata-rata tidak logis melebihi batas maksimal nilai. Jika ingin menjumlahkan seluruh nilai terlebih dahulu sebelum dibagi dengan 3, kode harus diperbaiki menggunakan tanda kurung ()

Program yang telah diperbaiki:

```python
nilai_uts = 80
nilai_uas = 90
nilai_tugas = 70

rata_rata = (nilai_uts + nilai_uas + nilai_tugas) / 3

print("Rata-rata:", rata_rata)
```

Pada kode yang telah diperbaiki, tanda kurung memaksa Python mengevaluasi penjumlahan seluruh nilai terlebih dahulu, sehingga pembagian dilakukan terhadap total nilai dan menghasilkan rata-rata yang benar.
