```python
# 1. Mengubah variabel total detik agar menerima masukan dari keyboard

total_detik = int(input("Masukkan total detik: "))

# Konversi Waktu
jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

# 2. Konversinal nilai 358200
print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
# 1. Modifikasi agar nilai mahasiswa dan standard kelulusan bisa menerima input keyboard
# 2. Menambahkan 5 nilai variabel tambahan dengan input keyboard

partisipasi = float(input("Masukkan nilai partisipasi: "))
tugas = float(input("Masukkan nilai Tugas: "))
kuis   = float(input("Masukkan nilai Kuis: "))
uts    = float(input("Masukkan nilai UTS: "))
uas    = float(input("Masukkan nilai UAS: "))

nilai_akhir = (0.30 * tugas) + (0.20 * partisipasi) + (0.10 * kuis) + (0.20 * uts) + (0.20 * uas)

standar_kelulusan = float(input("Masukkan nilai Standar Kelulusan: "))

apakah_lulus = nilai_akhir >= standar_kelulusan
perlu_remedial = nilai_akhir < standar_kelulusan

print(f"Nilai Akhir Mahasiswa: {nilai_akhir:.2f}")
print(f"Standar Kelulusan: {standar_kelulusan:.2f}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
# 1. Modifikasi inputan via keyboard
# 2. Modifikasi Diskon 35% jika belanja lebih dari 1.000.000

belanja = float(input("Masukkan total belanja: "))

diskon_awal = (0, 0.1 * belanja)[belanja > 500000]
diskon_tambahan = (0, 0.25 * belanja)[belanja > 1000000]

diskon = diskon_awal + diskon_tambahan

total_harga = belanja - diskon

print("Potongan Diskon =", diskon)
print("Total Harga Setelah Diskon =", total_harga)
```
