```python
# Soal 1 : Belanja Buku

jumlah_buku = 7
harga_per_buku = 18500
persen_diskon = 0.10
uang_budi = 150000
jumlah_orang = 4

total = jumlah_buku * harga_per_buku
diskon = int(total * persen_diskon)
bayar = total - diskon

kembalian = uang_budi - bayar
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang

uang_cukup = uang_budi >= bayar

print(f"Total       : {total}")
print(f"Diskon      : {diskon}")
print(f"Bayar       : {bayar}")
print(f"Kembalian   : {kembalian}")
print(f"Per orang   : {per_orang}")
print(f"Sisa        : {sisa}")
print(f"Uang cukup? : {uang_cukup}")
```

```python
# Soal 2 : PLTS Atap Rumah

jumlah_panel = 12
daya_panel_wp = 450
jam_penyinaran = 5
efisiensi = 0.80
kebutuhan_harian = 25
tarif_per_kwh = 1444.70

energi_harian = (jumlah_panel * daya_panel_wp * jam_penyinaran * efisiensi) / 1000
energi_bulanan = energi_harian * 30

nilai_listrik = energi_bulanan * tarif_per_kwh

kekurangan_harian = kebutuhan_harian - energi_harian
mencukupi = energi_harian >= kebutuhan_harian

print(f"Energi harian (kWh) : {round(energi_harian, 1)}")
print(f"Energi bulanan (kWh): {round(energi_bulanan, 1)}")
print(f"Nilai listrik (Rp)  : {round(nilai_listrik, 1)}")
print(f"Kekurangan (kWh)    : {round(kekurangan_harian, 1)}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
# Soal 3 : Generator Diesel dan Rugi Kabel

daya_generator = 200
jam_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

energi = daya_generator * jam_operasi
solar_harian = energi * konsumsi_solar
biaya = solar_harian * harga_solar

tangki_cukup = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

rugi_kabel = (arus ** 2) * hambatan
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

print(f"Energi (kWh)    : {energi}")
print(f"Solar (liter)   : {solar_harian}")
print(f"Biaya (Rp)      : {biaya}")
print(f"Tangki cukup    : {tangki_cukup} hari")
print(f"Sisa solar (L)  : {sisa_solar}")
print(f"Rugi kabel (W)  : {rugi_kabel}")
print(f"Rugi tinggi?    : {rugi_tinggi}")
```
