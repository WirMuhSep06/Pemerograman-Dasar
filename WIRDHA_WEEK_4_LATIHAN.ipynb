```python
# Menentukan Tahun Kabisat

tahun = int(input("Masukkan tahun: "))

tahun_kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

print(f"Apakah tahun {tahun} merupakan tahun kabisat? {tahun_kabisat}")
```

```python
# Menghitung Total Biaya Pemakaian Energi Listrik

daya_watt = float(input("Masukkan daya alat listrik: "))
lama_jam = float(input("Masukkan lama pemakaian per hari: "))
tarif_perkwh = float(input("Masukkan tarif listrik per kWh: "))

energi_kwh = (daya_watt * lama_jam) / 1000
biaya_harian = energi_kwh * tarif_perkwh
biaya_bulanan = biaya_harian * 30

print(f"Total Energi per Hari : {energi_kwh:.2f}")
print(f"Biaya Listrik per Hari: Rp {biaya_harian:,.3f}")
print(f"Biaya Listrik 30 Hari : Rp {biaya_bulanan:,.3f}")
```
