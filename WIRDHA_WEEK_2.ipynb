```python
# Jawaban Latihan 2 - Tipe primitif & type()

resistor = 1000
kapasitor = 0.00005
komponen = "Resistor"
Tersedia = True

print(resistor, type(resistor))
print(kapasitor, type(kapasitor))
print(Tersedia, type(Tersedia))
print(komponen, type(komponen))
```

```python
# Jawaban Latihan 3 - Memilih tipe yang tepat

nama_unit = "PLTS Nunukan"
jumlah_panel = 240
tegangan_keluaran = 48.5
sedang_beroperasi = True

print(nama_unit, type(nama_unit))
print(jumlah_panel, type(jumlah_panel))
print(tegangan_keluaran, type(tegangan_keluaran))
print(sedang_beroperasi, type(sedang_beroperasi))
```

```python
# Jawaban Latihan 4 -List

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Pembacaan pertama dan terakhir
print("Data pertama:", tegangan_harian[0])
print("Data terakhir:", tegangan_harian[5])

# (b) Menambahkan pembacaan baru
tegangan_harian.append(222.0)

# (c) jumlah total data
print("Jumlah data:", len(tegangan_harian))
```

```python
# Jawaban Latihan 5 - Dictionary

motor_listrik = {"nama": "Motor Induksi Industri",
                 "daya_kW": 8,
                 "tegangan_V": 380,
                 "rpm": 1450}

print("Daya motor:", motor_listrik["daya_kW"])

motor_listrik ["rpm"] = 1500

print(motor_listrik)
```

```python
# Jawaban Latihan 6 - Gabungan
pembangkit = [
    {"nama": "PLTGU", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "PLTA", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "PLTU", "kapasitas_MW": 80, "status": "pemeliharaan"},
]


print("Nama unit kedua:", pembangkit[1]["nama"])


pembangkit[0]["status"] = "pemeliharaan"


pembangkit.append({
    "nama": "PLTP",
    "kapasitas_MW": 120,
    "status": "operasi"
})


print(pembangkit)
```
